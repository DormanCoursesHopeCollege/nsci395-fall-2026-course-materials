## Four parts that are easy to confuse

| Part | What it holds | What Shift+Enter does |
|---|---|---|
| Markdown cell | Explanations, headings, lists, links, and your written answers | Renders the source as formatted text and selects the next cell |
| Code cell | Python instructions | Sends the code to the kernel and shows any result below |
| Output | Text, a table, an error, or a figure produced by code | Output is produced by a code cell; it is not a separate editable cell type |
| Kernel | Temporary running memory shared by the code cells | Keeps names and values until they are replaced or the kernel restarts |

The notebook file stores cell text and order. The kernel stores what has actually run. **Those two states can disagree!**

## Two modes: are you editing text or acting on a cell?

JupyterLab uses the same keyboard for two different jobs:

| Mode | How to enter it | What typing does |
|---|---|---|
| **Edit mode** | Select a cell and press **Enter** | Edits text at the cursor |
| **Command mode** | Press **Esc** | Applies notebook shortcuts to the selected cell |

In JupyterLab's default theme, an edit-mode cell has a blue border and a command-mode cell has a gray border. The more reliable clue is the text cursor: if it is blinking inside the cell, you are editing. Try `Esc`, then `Enter`, then `Esc` on this cell before moving on.

Six shortcuts are enough for today:

| Keys | Result |
|---|---|
| `Enter` | enter edit mode |
| `Esc` | return to command mode |
| `Shift+Enter` | run code or render Markdown, then select the next cell |
| `M` / `Y` in command mode | change the selected cell to Markdown / code |
| `Ctrl+S` (`Cmd+S` on macOS) | save |
| `Ctrl+Shift+C` (`Cmd+Shift+C` on macOS) | open the Command Palette |

The **Command Palette** is the searchable way to find an action when you do not know its menu or shortcut. Open it from **View → Command Palette** and search for `Show Keyboard Shortcuts…` and `Restart Kernel and Run All Cells…`; look at the results, but do not restart yet. If a shortcut ever types a letter into a cell, press `Esc` first. You were still in edit mode.

## The small Jupyter routine

1. **Select** a cell by clicking it.
2. Press **Enter** to edit; press **Esc** when you want shortcuts to act on the cell.
3. **Run or render** with Shift+Enter. A code cell executes; a Markdown cell renders.
4. **Add** a cell with +, set its type with the menu or command-mode `M`/`Y`, and move it with the toolbar arrows if its explanation is in the wrong place.
5. **Save** with the toolbar button or Ctrl/Cmd+S.
6. Use the **Command Palette** when you need to find an action. Use **Restart Kernel and Run All Cells** for a clean top-to-bottom check.

You do not need to memorize the full shortcut list. You do need to notice your mode, know what Shift+Enter will do to the selected cell type, and know how to find an unfamiliar command.

## Code cells have an execution history

A code cell shows brackets such as `[ ]`, `[*]`, or `[3]` to its left:

- `[ ]` means it has not run in the *current* kernel;
- `[*]` means the kernel is busy;
- a number records when that cell ran relative to other code cells.

The numbers need not increase from top to bottom. You can click a lower cell and run it first. That flexibility is useful, but it can also create output that the visible top-to-bottom notebook no longer explains.

# Reference: a small Markdown toolkit

Double-click a rendered Markdown cell to see its source. Press Shift+Enter to render it again.

| Source you type | Rendered role |
|---|---|
| `## Heading` | section heading |
| `- one item` | bullet |
| `1. first step` | numbered step |
| `**important**` | bold emphasis |
| `` `voltage_mV` `` | inline code or a variable name |
| `[label](https://example.org)` | link |

Markdown is plain text stored in the notebook. Rendering changes how it is displayed, not the underlying words. Use Markdown for predictions, interpretations, units, and notes. Use code cells only for instructions the kernel should execute.

To make a new prose cell in JupyterLab: add a cell, choose Markdown from the cell-type menu (or press `Esc`, then `M`), press `Enter` to edit, and press `Shift+Enter` to render. Move it with the toolbar arrows if its explanation belongs elsewhere.

### More command-mode shortcuts

Press `Esc` first so the shortcut acts on the selected cell instead of typing into it:

| Keys | Whole-cell action |
|---|---|
| Up Arrow / Down Arrow | select the cell above or below |
| `A` / `B` | insert a new cell above or below |
| `X` | cut the selected cell |
| `C` | copy the selected cell |
| `V` | paste the cut or copied cell below |

These actions affect an entire cell, including its type and contents. Use the Command Palette or the Edit menu if you do not remember a shortcut.

## Reference: notebook file versus kernel state

| Action | Cell text/order | Existing output | Kernel names/values |
|---|---|---|---|
| Edit a cell | changes | unchanged until execution | unchanged |
| Save | written to the notebook file | may be stored in an ordinary personal copy | unchanged |
| Run a code cell | unchanged unless you edited it | replaced for that cell | updated by that code |
| Restart kernel | unchanged | may remain visible until rerun or cleared | erased |
| Restart and Run All | unchanged | regenerated top to bottom | rebuilt top to bottom |

When a notebook behaves mysteriously, save it, restart the kernel, run all cells from the top, and stop at the first error or surprising result. This tests whether the visible notebook contains the steps needed to recreate its own result. It cannot detect a scientifically false label or unsupported claim by itself.
