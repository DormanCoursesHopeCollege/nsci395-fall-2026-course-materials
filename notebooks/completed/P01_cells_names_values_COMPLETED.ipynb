## Names refer to current values

![variable assignment illustration](https://d8it4huxumps7.cloudfront.net/uploads/images/65b8ebee43d04_python_variables_artboard_3.jpg?d=2000x2000)

In Python, a **statement** is a complete instruction. An **assignment statement** has the form:

```text
name = expression
```

Read the single equals sign `=` as “assign the value on the right to the name on the left.” It is not a permanent mathematical equality. A **later assignment can make the same name refer to a different value.** A **name** (also commonly called a variable) is a label Python uses to find a current **value**. An **expression** is code that Python evaluates to produce a value.

The examples `20` and `0.5` are numeric **literals**, meaning values written directly in the code. `20` is an integer (`int`), while `0.5` is a floating-point number (`float`).

We typically use lowercase names with underscores. A unit suffix such as `_s`, `_ms`, `_mV`, or `_hz` makes a scientific calculation easier to inspect. Python does not understand the suffix as a unit; it is a readability aid, not a unit check, so the code author must make sure the actual units are correct.

*Python names may contain uppercase or lowercase letters, digits, and underscores. They cannot contain spaces or begin with a digit, and a few words reserved for Python's own syntax cannot be names. Python is case-sensitive, so `rate_hz` and `Rate_hz` are different names. Course examples use lowercase names except where a later lesson explains another convention.*

### Live coding: assign values and evaluate an expression

Python writes addition, subtraction, multiplication, and division as `+`, `-`, `*`, and `/`. The symbols are **arithmetic operators**: they tell Python which calculation to perform. Parentheses group a calculation; otherwise, multiplication and division happen before addition and subtraction. A minus sign can also mark a negative number, as in `-70`.

For each assignment statement below, Python evaluates the expression on the right of `=` first and then assigns the resulting value to the name on the left. Here, `spike_count / window_s` means spike count divided by seconds. Python's `/` produces a floating-point result for these numeric values.

```python
spike_count = 20
window_s = 0.5
firing_rate_hz = spike_count / window_s
```

In the code cell below, a line beginning with `#` is a **comment** for a human reader. Python ignores the `#` and the rest of that line when it runs the cell.

````python
spike_count = 20
window_s = 0.5
firing_rate_hz = spike_count / window_s
````

## Known-answer check

Twenty spikes in half a second must be 40 spikes per second. An *assertion* records that expectation. If the expression or units change incorrectly, the assertion should fail instead of letting a plausible-looking number pass unnoticed.

The Python `assert` statement checks a **condition**. In this lesson, each asserted condition is a comparison that evaluates to one of the Boolean (`bool`) values `True` or `False`. If the condition is `True`, execution continues without output. If it is `False`, Python raises an `AssertionError` to make the failed expectation visible.

Notice the double equals sign `==`. This **comparison operator** asks whether its left and right values are equal and evaluates to `True` or `False`. Unlike the single `=` used for assignment, `==` does not change a name's value.

`print(firing_rate_hz)` is a **function call**. A function is reusable code that performs an operation; writing `function(argument)` calls it, and the value inside the parentheses is its **argument**. Here, `print(...)` displays the value without changing it.

````python
assert firing_rate_hz == 40
print(firing_rate_hz)
````

## Put the result in a readable sentence

An f-string (short for “formatted string”) combines fixed text with current values:

| Part | Meaning |
|---|---|
| `f'...'` | Make a *f*ormatted string |
| `{firing_rate_hz}` | Insert the current value |
| `:.1f` | Use fixed-point format with one digit after the decimal point |
| `Hz` | A text label that we supply |

The formatting does not add or check units. The code author remains responsible for calculating in seconds and writing the correct unit.

### Live coding: report the value and unit

Point to the Python name, the formatting instruction, and the unit text before running the line.

```python
print(f'Firing rate: {firing_rate_hz:.1f} Hz')
```

A **string** (`str`) is text, which can contain letters, numbers, symbols, and spaces. A **string literal**, meaning text written directly in code, uses matching single or double quotation marks: `'my string'` or `"my string"`, but not `"my string'`. The leading `f` makes the quoted text above an f-string.

````python
print(f'Firing rate: {firing_rate_hz:.1f} Hz')
````

These lines change the spike count and then display both names. They do not repeat the division. Inside `print(...)`, a comma separates arguments. For example, `print('spike_count:', spike_count)` gives `print` two values to display in order, with a space between them.


````python
spike_count = 24
print('spike_count:', spike_count)
print('firing_rate_hz:', firing_rate_hz)
````

The two values are now inconsistent: `spike_count` is 24, while `firing_rate_hz` remains 40. Python evaluated the earlier division once. The result does not remember where it came from and does not update like a spreadsheet formula.

This is a **stale value**. It is valid Python and may look reasonable, which makes it more dangerous than an obvious syntax error.

### Live coding: recompute the dependent value

```python

print(f'Updated firing rate: {firing_rate_hz:.1f} Hz')
```

## Text must become a number before arithmetic

When these types are printed, Python displays `<class 'str'>` and `<class 'int'>`; for this lesson, read those as “string type” and “integer type.”

Function calls can be nested inside other calls. In `print(type(x), type(y))`, Python evaluates the two inner `type(...)` calls first. The comma then separates the two results given as arguments to `print(...)`. ## Reference: the few value types used today

| Example | Type | Use today |
|---|---|---|
| `20` | integer (`int`) | whole spike count |
| `0.5` | floating-point number (`float`) | duration or measured value |
| `'20'` | string (`str`) | numeric text that must be converted before arithmetic |
| `'Hz'` | string (`str`) | a unit label shown to a reader |
| `True` or `False` | Boolean (`bool`) | used here for comparisons and assertions; studied further in P05 |

Use `type(value)` to inspect a value's type. For example, Python displays `type('9')` as `<class 'str'>`; for now, read that as “the string type.” `int('20')` performs a **type conversion** from numeric text to an integer. Dividing a millisecond value by 1000 performs a separate **unit conversion**; dividing count by seconds is the **calculation**; and an f-string controls **presentation**. The expression inside `{...}` is evaluated before insertion, while `:.1f` changes display precision rather than the stored value. Writing `Hz` beside a number does not convert milliseconds to seconds.

Useful name rules:

- begin with a letter;
- use letters, digits, and underscores;
- do not include a unit that the value does not actually use;
- remember that `rate_hz` and `Rate_hz` are different names.

### Syntax used in P01

| Syntax | Meaning |
|---|---|
| `# explanation` | Start a comment; Python ignores the rest of that line. |
| `...` in explanatory text | Mark where an actual value or expression belongs; do not copy the three dots in P01. |
| `name = expression` | Evaluate the right-hand expression, then assign its value to the name. |
| `20`, `0.5`, `'9'` | Write integer, floating-point, and string literals directly in code. |
| `+`, `-`, `*`, `/` | Add, subtract (or mark a negative number), multiply, and divide. |
| `(expression)` | Group an arithmetic expression so it is evaluated first. |
| `function(argument)` | Call a function with one argument. |
| `print(first, second)` | Use a comma to give `print` two arguments. |
| `left == right` | Compare two values for equality; the result is `True` or `False`. |
| `assert condition` | Continue if the condition is `True`; raise `AssertionError` if it is `False`. |
| `f'Text {value:.1f}'` | Insert a value into text and show one digit after the decimal point. |
| `int(text)` | Convert numeric text to a new integer value. |
| `type(value)` | Inspect the type of a value. |

For ungrouped arithmetic, Python follows the usual precedence: multiplication and division before addition and subtraction. Parentheses make the intended order explicit.
