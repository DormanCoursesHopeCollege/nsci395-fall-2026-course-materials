## Scientific setting

We will use small **synthetic observations**. Each observation supplies a spike count from one neuron and the duration of one counting window. A rate in hertz means spikes per second:

$$
\text{rate (Hz)} = \frac{\text{spike count}}{\text{window (s)}}
$$

The examples teach calculation and reporting. They do not represent trial variability, measurement noise, or a population of neurons, so they cannot support a biological claim about a real neuron.

## Expressions carry values and units

Python evaluates the right side of `=` and stores the result under the name on the left. Division produces a floating-point value here, even when the spike count and millisecond window are whole numbers.

A suffix such as `_ms`, `_s`, or `_hz` is a promise from the programmer, not a unit system enforced by Python. The conversion must still appear in the expression:

$$
\text{window\_s} = \frac{\text{window\_ms}}{1000}
$$

### Worked step 1: convert the window and calculate rate

````python
spike_count = 15
window_ms = 250
window_s = window_ms / 1000
rate_hz = spike_count / window_s
print(window_s, rate_hz)
````

## Known-answer check

A check compares the computed values with expectations justified before coding. It protects this example from a missing conversion or wrong denominator. It does not prove that every future rate calculation is correct.

`assert` checks whether a condition is `True`. The operator inside the condition determines what is compared:

- `==` asks whether two values are equal, as in `rate_hz == 60`.
- `is` asks whether two expressions refer to the very same Python object, not merely equal values. In `type(window_s) is float`, Python checks that `type(window_s)` is exactly the `float` type object.

Use `==` for ordinary number or string comparisons. Do not write `window_s is 0.25`; `is` is used here only for the type checks.

````python
assert window_s == 0.25
assert rate_hz == 60
assert type(window_s) is float
assert type(rate_hz) is float
print('Rate and type checks passed.')
````

## Numeric values, strings, and f-strings

Python uses different value types for different operations:

| Example | Type | Role here |
|---|---|---|
| `15` | `int` | whole spike count |
| `0.25` | `float` | duration or rate calculation |
| `'15'` | `str` | text that might arrive from a form or file |

`int('15')` converts numeric text to an integer. That is a **type conversion**. Dividing milliseconds by 1000 is a separate **unit conversion**.

An f-string inserts evaluated values inside braces. `:.1f` changes display precision, not the stored number, and writing `Hz` does not repair a calculation performed in spikes/ms.

### Worked step 2: convert text and report the result

# Reference: expressions, types, and units

An expression combines literal values, current name values, operators, or function calls to produce one value. Python uses parentheses first, then multiplication and division, then addition and subtraction. Parentheses are worth adding when they make the scientific relationship visible.

| Expression | Result | Scientific role |
|---|---:|---|
| `250 / 1000` | `0.25` | convert ms to s |
| `15 / 0.25` | `60.0` | spikes per second |
| `40 - 12` | `28` | absolute rate change in Hz |
| `100 * (40 - 12) / 12` | about `233.3` | percent change relative to baseline |

In a percent change, the baseline belongs in the denominator. A percent is dimensionless, while the absolute difference remains in hertz. Percent change is follow-up material here, not a prerequisite for P03.

## Reference: strings and readable output

A string is text between quotes. Joining two strings with `+` concatenates text; adding two numbers performs arithmetic. Python refuses some mixed operations because their meaning is ambiguous:

```python
'9' + '1'       # '91'
9 + 1           # 10
int('9') + 1    # 10
```

In an f-string, braces contain expressions evaluated at runtime. A format such as `{rate_hz:.1f}` requests one digit after the decimal point. Formatting does not change the stored value and does not validate its unit.

## Reference: a plausible wrong result

`15 / 250` produces `0.06`. Python has done exactly what it was asked to do: the unit is spikes/ms. Printing `0.06 Hz` creates a silent scientific error that is too small by a factor of 1000.

A useful diagnosis asks four questions:

1. What does each value represent?
2. What is each unit before the operation?
3. What unit should the result have?
4. Is there a simple case whose answer is already known?
